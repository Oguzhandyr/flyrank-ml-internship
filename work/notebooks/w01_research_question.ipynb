## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

* **Search Question:** Which high-visibility pages are currently showing early signals of traffic decay, and what specific content refresh actions should be prioritized?
* **Unit of Analysis:** A single indexed URL (page level) observed over a 90-day window.
* **The Output:** A ranked priority score (0.0 to 1.0) along with an action flag (e.g., `PRIORITIZE_REFRESH`, `MONITOR`, `NO_ACTION`).
* **The Human Action:** SEO and content marketing teams allocate writer and editor hours directly to the top-K flagged pages instead of manually auditing tens of thousands of URLs.
* **Cost of a Wrong Call:**
  * *False Positive (Flagging a healthy page):* Content team wastes hours rewriting an already performing article (opportunity cost).
  * *False Negative (Missing a dying page):* The page loses rank and organic traffic permanently, resulting in direct revenue loss.
* **Why ML/Data is Needed:** Manual review across 30,000+ client URLs is infeasible. Simple heuristic thresholds fail to balance age vs. impression dynamics, whereas a tree-based model captures non-linear interactions across visibility, staleness, and click-through rates.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os
import subprocess
import pandas as pd
import numpy as np

REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if "google.colab" in str(get_ipython()):
    if not os.path.exists(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    DATA_PATH = os.path.join(REPO_DIR, "data/raw/content_refresh_anonymized.csv")
else:
    DATA_PATH = "../../data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(DATA_PATH)

total_pages = len(df)
declining_rate = df['trend_direction'].str.lower().eq('down').mean()

stale_visible = df[(df['days_since_last_update'] >= 180) & (df['impressions_90d'] >= 500)]
stale_visible_declining = stale_visible['trend_direction'].str.lower().eq('down').mean()

median_impressions = df['impressions_90d'].median()
median_age = df['days_since_last_update'].median()

print(f"Total Pages Analyzed: {total_pages}")
print(f"Overall Declining Rate (Baseline): {declining_rate:.1%}")
print(f"Stale & High-Visibility Pages Count: {len(stale_visible)}")
print(f"Declining Rate in Stale/Visible Segment: {stale_visible_declining:.1%}")
print(f"Median Impressions (90d): {median_impressions:.1f} | Median Days Since Update: {median_age:.1f} days")
```
